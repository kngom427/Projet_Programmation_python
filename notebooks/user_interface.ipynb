<span style="color:yellow;"> 1.3 Création du corpus et ajout des document </span>

```python
# -  Instanciation d'un objet corpus avec un theme DiscoursUS
corpus = Corpus("DiscoursUS")

# definition de la fonction de découpage en phrases
def split_into_sentences(text):
    # Découpage simple en phrases
    sentences = re.split(r'[.!?]', text)
    return [s.strip() for s in sentences if len(s.strip()) > 20]

# - Ajout des documents (chaque phrase devient un Document)
for _, row in df.iterrows():
    auteur = row["speaker"]   
    titre = row["descr"]
    date = row["date"]
    url = row["link"]
    texte = row["text"]

     # - Découpage en phrases
    phrases = split_into_sentences(texte)

    for phrase in phrases:
        
        # Verifier que la phrase n'est pas vide avant de l'ajouter
        if phrase.strip():
            corpus.add_document(titre, auteur, date, url, phrase.strip())
```

```python
print("--------- Corpus créé : ---------")
print(corpus)
```

### <u style="color: red;"> Partie 3:</u>   Petite interface


* Champ de saisie de la requête

```python
input_requete = widgets.Text(
    value='',
    placeholder='Entrez votre requête (ex: freedom democracy)',
    description='Requête :',
    layout=widgets.Layout(width='80%')
)
```

* Curseur pour le nombre de documents (k)

```python
slider_k = widgets.IntSlider(
    value=5,
    min=1,
    max=20,
    step=1,
    description='Top k :',
    continuous_update=False
)
```

* Bouton de recherche

```python
bouton_recherche = widgets.Button(
    description='Lancer la recherche',
    button_style='primary',
    icon='search'
)
```

* Zone d’affichage des résultats

```python
zone_resultats = widgets.Output()
```

* Fonction déclenchée au clic du bouton

```python
def lancer_recherche(b):
    with zone_resultats:
        clear_output()
        
        requete = input_requete.value
        k = slider_k.value
        
        if requete.strip() == "":
            print(" Veuillez entrer une requête.")
            return
        
        print(f" Recherche pour : '{requete}' (top {k})\n")
        
        df_resultats = moteur_recherche.search(requete, k)
        display(df_resultats)
```

* Lier le bouton à la fonction

```python
bouton_recherche.on_click(lancer_recherche)
```

* Affichage final de l’interface

```python
display(
    widgets.VBox([
        input_requete,
        slider_k,
        bouton_recherche,
        zone_resultats
    ])
)
```
